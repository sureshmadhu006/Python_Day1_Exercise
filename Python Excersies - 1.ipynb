```python
# Write a program to find whether the given year is leap year or not ?
```

```python
y=int(input("Enter the year: "))
if (y%4==0 and y%100!=0) or (y%400==0):
    print("The year is leap year")
else:
    print("The year is not a leap year")
    
```
